```python
import os
from lxml import etree

def merge_gpx_files(folder_path, output_file):
    gpx_ns = "http://www.topografix.com/GPX/1/1"
    nsmap = {None: gpx_ns}

    # Create the root element for the merged GPX file
    root = etree.Element("gpx", version="1.1", creator="GPX Merger", nsmap=nsmap)
    merged_trk = etree.SubElement(root, "trk")
    merged_trkseg = etree.SubElement(merged_trk, "trkseg")

    # Get all GPX files, sort them numerically and iterate
    gpx_files = [f for f in os.listdir(folder_path) if f.lower().endswith(".gpx")]
    gpx_files.sort(key=lambda x: int(''.join(filter(str.isdigit, x[:1])) or 0))  # sort by first digit

    for file_name in gpx_files:
        file_path = os.path.join(folder_path, file_name)
        tree = etree.parse(file_path)
        trksegs = tree.findall(".//{%s}trkseg" % gpx_ns)

        for seg in trksegs:
            for trkpt in seg.findall("{%s}trkpt" % gpx_ns):
                merged_trkseg.append(trkpt)

    # Write the merged GPX data to the output file
    tree = etree.ElementTree(root)
    tree.write(output_file, xml_declaration=True, encoding="UTF-8", pretty_print=True)
    print(f"Merged GPX saved to: {output_file}")

# Example usage
merge_gpx_files("/Users/adikteev/Downloads/GPX to merge", "merged_output.gpx")
```
